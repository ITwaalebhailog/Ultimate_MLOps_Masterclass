```python
#Palindrome

def is_palidrome(text):
    reversed_text = text[::-1]

    if text == reversed_text:
        return True
    else:
        return False

print(is_palidrome("radar"))
result = is_palidrome("python")
print(result)
```

```python
#Find the largest number without using min and max or in built in fucntions
numbers = [1,2,3,4,5,100,4,90]


print(len(numbers))

def find_largest(numbers):
    if len(numbers) == 0:
        return None
    largest = numbers[0]  

    for num in numbers:  
        if num > largest:
            largest = num
    return largest

print(find_largest(numbers)) 
```

```python
# Find the second largest number without using min, max, or built-in functions
numbers = [1, 2, 3, 4, 5, 100, 4, 90, 101, 103]


def find_second_largest(numbers):
    if len(numbers) < 2:
        return None

    largest = None
    second_largest = None

    for num in numbers:                                        
        if largest is None or num > largest:
            second_largest = largest
            largest = num
        elif num != largest:
            if second_largest is None or num > second_largest:
                second_largest = num

    return second_largest


print(find_second_largest(numbers))
```

```python
# frequency of numbers

data = [2,5,4,6,7,8,9,10,1,2,3,4,5,1,1,1,1,1,1]

def count_occur(numbers, target):
    tally = 0
    for num in numbers:
        if num == target:
            tally = tally + 1
    return tally

print(count_occur(data, 1))
```

```python
#Remove Duplicates

def remove_duplicates(items):
    unique_items = []

    for item in items:
        if item not in unique_items:
            unique_items.append(item)
    return unique_items

raw_list = [1,1,1,1,2,3,4,5,6,6,88,90,9,76,5]
print(remove_duplicates(raw_list))
```

```python
#character frequency

def count_frequencies(char_list):
    freq_map = {}

    for char in char_list:
        if char in freq_map:
            freq_map[char] = freq_map[char] + 1
        else:
            freq_map[char] = 1
    return freq_map

chars = ['a','b','a','b','c','d','c']
print(count_frequencies(chars))
```
