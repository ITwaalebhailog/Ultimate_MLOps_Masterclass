```python
def split_bill():
    try:
        total_bill = float(input("Enter total bill amount: "))
        people = int(input("Enter number of people: "))
        
        share = total_bill / people
        print(f"each person pays: ${share:.2f}")

    except ValueError:
        print("[Error] Invalid input. please enter numeric float values")

    except ZeroDivisionError:
        print("[Error] Number of people cannot be zero. Please enter a valid number.")


split_bill()

        
```

```python
 def register_user(username, age):
    try:
        print("-----------> start registeration process <------------")
        if len(username) < 3:
            raise ValueError("Username must be at least 3 characters long.")
        if age < 18:
            raise ValueError("User must be at least 18 years old to register.")

    except ValueError as error:
        print(f"[Registration Rejected]: error {error}")

    else:
        print(f"[Success]: User '{username}' registered successfully.")

    finally:
        print("[System]: Registration session closed.")

register_user("ali", 17)
```

```python
class InsufficientFundsError(Exception):
    """Custom exception for insufficient funds in a bank account."""
    pass

def withdraw_money(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(f"Insufficient funds: Cannot withdraw ${amount}. Current balance is ${balance}.")
    balance -= amount
    print(f"Withdrawal successful. New balance: ${balance:.2f}")
    return balance

try:
    withdraw_money(balance=100, amount=150)
except InsufficientFundsError as e:
    print(f"[Error]: {e}")

```
