```python
score = int(input("Enter you score"))

if score >= 50:
    print("status: Passed")
else:
    print("status: Fails")
```

```python
temperature = int(input("Enter the temperature"))

if temperature > 35:
    print("It's a hot day")
elif temperature >= 20:
    print("It's a nice day")
elif temperature >= 10:
    print("It's a bit cold")
else:
    print("It's a cold day")
```

```python
age = int(input("Enter your age"))
if age >= 18:
    has_license = input("Do you have a driving license? (yes/no): ").upper()
    if has_license == "YES":
        print("You can drive.")
    else:
        print("You cannot drive without a license.")
else:
    print("You are too young to drive.")
```

```python
for num in range(1, 7):
    if num % 2 == 0:
        print(num, "is even")
    else:
        print(num, "is odd")
```

```python
services = ["nginx", "apache", "mysql", "redis"]
for app in services:
    if app == "nginx" or app == "apache":
        print(app, "is a web server")
    else:
        print(app, "is not a web server")
```

```python
teams = ["Team Alpha", "Team Beta"]
scores = [92, 45]

for team in teams:
    print("teams:", team)
    for score in scores:
        if score >= 50:
            if score >= 90:
                print(f"Score {score}: outstanding performance")
            else:
                print(f"Score {score}: standard passed")

        else:
            print(f"Score {score}: Needs improvement")
```
