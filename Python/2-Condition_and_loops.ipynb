```python
temperature = int(input("Enter the temperature"))

if temperature > 35:
    print("It's a hot day")
elif temperature >= 20:
    print("It's a nice day")
elif temperature >= 10:
    print("It's a bit cold")
else:
    print("It's a cold day")
```

```python
count = 5

while count > 0:
    if count == 1:
        print("1.. Final second!")
    else:
        print(count, "seconds remaining...")
    count=count - 1  #count -= 1
```

```python
balance = 100
item_cost = 25

while balance > 0:
    balance = balance - item_cost
    if balance > 25:
        print("Item purchased. Remaining balance:", balance)
    elif balance > 0:
        print("Item purchased. low balance:", balance)
    else:
        print("Item purchased. Insufficient balance:", balance)
```

```python
attempts = 3

while attempts > 0:
    pin = input("Enter your PIN: ")

    if pin == "1234":
        print("Access granted.")
        break
    else:
        attempts -= 1
        if attempts > 0:
            print("Incorrect PIN. You have", attempts, "attempts left.")
        else:
            print("Incorrect PIN. No attempts left. Access denied.")
```
