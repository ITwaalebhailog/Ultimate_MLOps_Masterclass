```python
with open("notes.txt", "w") as file:
    file.write("Python Lecture Day 5")
    file.write("Topic: File Handling")
    file.read(file)
    print(file)

print("File written successfully.")
```

```python
new_entries = [
    "Topic 1: File Handling\n",
    "Topic 2: Binary Files\n",
    "Topic 3: Exception Handling\n"
]

with open("notes.txt", "a") as file:
    file.writelines(new_entries)

print("Next lines appended successfully.")
```

```python
with open("notes.txt", "r") as file:
    print("----Reading file Content ---")
    for line in file:
        print("line:", line)
```

```python
raw_data = b'\x00\x01\x02\x03\x04\x05\x06\x07\x08\x09' 

with open("sample_image.bin", "wb") as bin_file:
    bin_file.write(raw_data)

print("Binary file created")


with open("sample_image.bin", "rb") as bin_file:
    content = bin_file.read()
    print("Raw Binary Content:", content)
    print(len(content), "bytes read from the binary file.")
```
