Data Structures:
List ---> [] --> [1,2,"mlops",3.15]--> ordered ---> mutable ---> Allows Duplicates
Tuple --> () --> (1,2,3,5) --> ordered --> immutable ---> Allows Duplicates
Set ---> {} or set() {1,2,3,4,5} --> unordered ---> mutable --> Dont Allow Duplicates
Dictionary --> {key:value} --> {"name":"JOhn","age":30} --> ordered (3.7+) --> mutable --> Keys(Unique(NO)):                                                                                          values (Dupliates(Yes)) 


```python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
fruits.append("orange")

print(fruits)

#index
print(fruits[0])
print(fruits[-1])

fruits.remove("cherry")
print(fruits)
```

```python
numbers = [10,25,30,45,50,65,70]

middle_nums = numbers[2:5]
print("Slice: ", middle_nums)

filtered = []
for n in numbers:
    if n > 40:
        filtered.append(n)

print("Above 40:", filtered)
```

```python
word = str(input("Enter your word")) #palindrome
if word == word[::-1]:
    print("Palindrome", word)
else:
    print("not a palindrome")
```

```python
raw_user_ids = [101,102,103,101,105,102,106]
unique_users = set(raw_user_ids)
print("Unique order list: ", unique_users)

unique_users.add(107)
unique_users.discard(101)
print("Modified Set: ", unique_users)
```

```python
dev_tools = {"git", "Docker", "Python", "Vscode"}
ops_tools = {"Docker", "Kubernetes", "terraform", "Python"}

print("All tools:", dev_tools | ops_tools)

print("Common Tools: ", dev_tools & ops_tools)

print("Dev only tools:", dev_tools - ops_tools)
```

```python
app_settings = {"Debug": True, "Version": "1.4.2"}

log_level = app_settings.get("log", "INFO")
print("Log level: ", log_level)

debug_mode = app_settings.pop("Version")
print("Removed :", debug_mode)
print("Settings:" , app_settings)
```
