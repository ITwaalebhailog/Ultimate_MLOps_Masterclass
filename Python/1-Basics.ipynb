Arithematic Operators

```python
a = 10
b = 20
print(a + b)
print("Division (float):", a / b)
print("Division (integer):", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
```

```python
marks = 75
print(marks >= 50)
print(marks <= 50)
print(marks == 100)
print(marks != 0)
```

```python
#and or not
age = 20
has_id = True

# and = every value has to match the condtion
# or = any value has to match the condition
# not = reverse the boolean value

can_enter = (age >= 18) and has_id
print("Entry allowed:", can_enter)

can_enter = (age >= 18) or has_id
print("Entry allowed:", can_enter)

is_blocked = False
print("Active User:", not is_blocked)
```
